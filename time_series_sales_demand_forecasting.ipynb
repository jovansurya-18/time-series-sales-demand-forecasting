**The Very Hungry Caterpillar**

A multiplicative decomposition model was selected for The Very Hungry Caterpillar because the magnitude of seasonal fluctuations increases as the overall sales level increases. The seasonal spikes become larger over time, indicating that the seasonal effect is proportional to the trend rather than constant.

The observed series shows strong recurring spikes in sales, likely associated with seasonal purchasing behaviour such as holidays, birthdays, and educational cycles. The trend component indicates a long-term upward movement in sales over the analysed period, suggesting sustained popularity and growth over time.

The seasonal component displays a clear repeating yearly pattern, demonstrating strong seasonality in weekly sales. **The Alchemist**

An additive decomposition model was selected for The Alchemist because the seasonal variation remains relatively stable over time. Although periodic spikes occur, their magnitude does not increase proportionally with the overall sales level.

The observed series shows recurring spikes and moderate fluctuations throughout the period. The trend component suggests gradual long-term growth with some periods of stability and decline. Compared with The Very Hungry Caterpillar, the overall growth pattern is less pronounced.

The seasonal component reveals repeating yearly patterns in sales, indicating consistent seasonal behaviour across years. The residual component captures irregular fluctuations and several sharp outliers that are not explained by the trend or seasonality, possibly due to temporary marketing campaigns or unusual demand shocks.

```python
# Plotting ACF and PACF on The Alchemist

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

plot_acf(
    alchemist_df["Volume"],
    lags=52,
    ax=axes[0]
)

plot_pacf(
    alchemist_df["Volume"],
    lags=52,
    ax=axes[1]
)

axes[0].set_title("ACF - The Alchemist")
axes[1].set_title("PACF - The Alchemist")

plt.show()
```

**The Very Hungry Caterpillar**

The ACF plot shows very strong positive autocorrelation that decays slowly across increasing lags. This indicates that the sales series has strong persistence and is likely non-stationary, meaning current sales are highly influenced by previous sales behaviour over long periods.

A noticeable increase in autocorrelation appears again near lag 52, which suggests yearly seasonality within the weekly sales data. This supports the earlier decomposition findings that identified recurring annual sales patterns.

The PACF plot shows one dominant spike at lag 1, followed by relatively small values for later lags. This indicates that the immediate previous week has the strongest direct effect on current sales, suggesting an autoregressive structure primarily driven by short-term dependence.

Overall, the results indicate strong temporal dependence, yearly seasonal behaviour, and a likely autoregressive process.

---

**The Alchemist**

The ACF plot also demonstrates positive autocorrelation, although the decay is faster and weaker compared with The Very Hungry Caterpillar. This suggests the series still contains persistence and trend behaviour, but with less long-term dependence.

```python
# Now for LSTM

# Original series
cat_series = y_caterpillar.values.reshape(-1,1)
alc_series = y_alchemist.values.reshape(-1,1)

# Split
cat_train = cat_series[:-32]
cat_test = cat_series[-32:]

alc_train = alc_series[:-32]
alc_test = alc_series[-32:]

# Scale
cat_scaler = MinMaxScaler()

cat_train_scaled = cat_scaler.fit_transform(cat_train)
cat_test_scaled = cat_scaler.transform(cat_test)

alc_scaler = MinMaxScaler()

alc_train_scaled = alc_scaler.fit_transform(alc_train)
alc_test_scaled = alc_scaler.transform(alc_test)
```

```python
# Create sequence function

def create_sequences(data, window_size):

    X = []
    y = []

    for i in range(window_size, len(data)):

        X.append(
            data[i-window_size:i]
        )

        y.append(
            data[i]
        )

    return np.array(X), np.array(y)
```

```python
# Create training sequence

window_size = 12

X_train_cat, y_train_cat = create_sequences(
    cat_train_scaled,
    window_size
)

X_train_alc, y_train_alc = create_sequences(
    alc_train_scaled,
    window_size
)
```

```python
# Create test sequence

cat_combined = np.concatenate([
    cat_train_scaled[-window_size:],
    cat_test_scaled
])

X_test_cat, y_test_cat = create_sequences(
    cat_combined,
    window_size
)

alc_combined = np.concatenate([
    alc_train_scaled[-window_size:],
    alc_test_scaled
])

X_test_alc, y_test_alc = create_sequences(
    alc_combined,
    window_size
)
```

```python
# Build tunable LSTM model

def build_model(hp):

    model = Sequential()

    model.add(
        LSTM(
            units=hp.Int(
                "units",
                min_value=32,
                max_value=128,
                step=32
            ),
            activation="tanh"
        )
    )

    model.add(
        Dropout(
            hp.Float(
                "dropout",
                min_value=0.0,
                max_value=0.5,
                step=0.1
            )
        )
    )

    model.add(Dense(1))

    model.compile(
        optimizer="adam",
        loss="mse"
    )

    return model
```

```python
# KerasTuner

import keras_tuner as kt

cat_tuner = kt.RandomSearch(
    build_model,
    objective='val_loss',
    max_trials=5,
    overwrite=True,
    directory='cat_tuning',
    project_name='lstm'
)

cat_tuner.search(
    X_train_cat,
    y_train_cat,
    epochs=20,
    validation_split=0.2,
    verbose=1,
    shuffle=False
)

alc_tuner = kt.RandomSearch(
    build_model,
    objective='val_loss',
    max_trials=5,
    overwrite=True,
    directory='alc_tuning',
    project_name='lstm'
)

alc_tuner.search(
    X_train_alc,
    y_train_alc,
    epochs=20,
    validation_split=0.2,
    verbose=1,
    shuffle=False
)
```

```python
# Get best model for each book

best_cat_model = cat_tuner.get_best_models(1)[0]
best_alc_model = alc_tuner.get_best_models(1)[0]
```

```python
# Scale the residuals

cat_res_scaler = MinMaxScaler()

cat_res_train = cat_residuals[:-32].values.reshape(-1,1)
cat_res_test = cat_residuals[-32:].values.reshape(-1,1)

cat_res_train_scaled = (cat_res_scaler.fit_transform(cat_res_train))
cat_res_test_scaled = (cat_res_scaler.transform(cat_res_test))


alc_res_scaler = MinMaxScaler()

alc_res_train = alc_residuals[:-32].values.reshape(-1,1)
alc_res_test = alc_residuals[-32:].values.reshape(-1,1)

alc_res_train_scaled = (alc_res_scaler.fit_transform(alc_res_train))
alc_res_test_scaled = (alc_res_scaler.transform(alc_res_test))
```

```python
# Create train sequences

window_size = 12

X_train_cat_res, y_train_cat_res = create_sequences(
    cat_res_train_scaled,
    window_size
)

X_train_alc_res, y_train_alc_res = create_sequences(
    alc_res_train_scaled,
    window_size
)
```

```python
# Create test sequences

cat_res_combined = np.concatenate([
    cat_res_train_scaled[-window_size:],
    cat_res_test_scaled
])

X_test_cat_res, y_test_cat_res = create_sequences(
    cat_res_combined,
    window_size
)

alc_res_combined = np.concatenate([
    alc_res_train_scaled[-window_size:],
    alc_res_test_scaled
])

X_test_alc_res, y_test_alc_res = create_sequences(
    alc_res_combined,
    window_size
)
```

```python
# KerasTuner on the residuals
# Reuse the previous build_model function

cat_res_tuner = kt.RandomSearch(
    build_model,
    objective="val_loss",
    max_trials=5,
    overwrite=True,
    directory="cat_hybrid",
    project_name="lstm_residuals"
)

cat_res_tuner.search(
    X_train_cat_res,
    y_train_cat_res,
    epochs=20,
    validation_split=0.2,
    verbose=1,
    shuffle=False
)

alc_res_tuner = kt.RandomSearch(
    build_model,
    objective='val_loss',
    max_trials=5,
    overwrite=True,
    directory='alc_hybrid',
    project_name='lstm_residuals'
)

alc_res_tuner.search(
    X_train_alc_res,
    y_train_alc_res,
    epochs=20,
    validation_split=0.2,
    verbose=1,
    shuffle=False
)
```

```python
# Get best hybrid model for each book

best_cat_res_model = (
    cat_res_tuner.get_best_models(1)[0]
)

best_alc_res_model = (
    alc_res_tuner.get_best_models(1)[0]
)
```

```python
# Now predcit the residuals with the best models

cat_res_pred_scaled = (
    best_cat_res_model.predict(
        X_test_cat_res
    )
)

alc_res_pred_scaled = (
    best_alc_res_model.predict(
        X_test_alc_res
    )
)
```

```python
# Converting the residuals back to the original value

cat_res_pred = (
    cat_res_scaler.inverse_transform(
        cat_res_pred_scaled
    )
)

alc_res_pred = (
    alc_res_scaler.inverse_transform(
        alc_res_pred_scaled
    )
)
```
